# Binary Search

**Binary Search** is an efficient algorithm used to find the position of a target value within a **sorted array**. It operates on the **divide-and-conquer** principle, repeatedly halving the search space until the element is located or determined to be absent.

---

## 2. How It Works

1. **Initialize Pointers:** Start with a lower pointer (`low`) at index `0` and an upper pointer (`high`) at the last index of the array (`n - 1`).

2. **Calculate Midpoint:** Find the middle index using the formula:
   $$
   \text{mid} = \text{low} + \frac{\text{high} - \text{low}}{2}
   $$
   *(Note: This specific layout avoids integer overflow bugs in languages like Java/C++).*

3. **Evaluate and Eliminate:** 
   * **If Target == Array[mid]:** The search is successful; return `mid`.
   * **If Target < Array[mid]:** The target can only be in the left half. Discard the right half by updating `high = mid - 1`.
   * **If Target > Array[mid]:** The target can only be in the right half. Discard the left half by updating `low = mid + 1`.

4. **Repeat:** Continue steps 2 and 3 until the target is found, or `low` becomes greater than `high` (meaning the target is not in the array).

---

## 3. Complexity Analysis

### Time Complexity
Every iteration of the algorithm eliminates **half of the remaining search space**. For an array of size n, it takes at most $\log_2(n)$ steps to reduce the array down to a single element.

* **Best Case: $\mathcal{O}(1)$** – The target element is located exactly at the first midpoint checked.
* **Average Case: $\mathcal{O}(\log n)$** – The target is found somewhere along the halving process.
* **Worst Case: $\mathcal{O}(\log n)$** – The target is at the extreme ends of the array, or does not exist in the collection at all.

### Space Complexity
The auxiliary memory footprint depends entirely on the design pattern used:

* **Iterative Approach: $\mathcal{O}(1)$** – Uses a single loop with constant auxiliary memory for index pointers (`low`, `high`, `mid`).
* **Recursive Approach: $\mathcal{O}(\log n)$** – Each split pushes a new function frame onto the system call stack, scaling logarithmically with the dataset size.

---

## 4. Binary Search vs. Traditional Search (Linear Search)

The traditional method—**Linear Search**—checks every item sequentially from beginning to end.

| Feature | Traditional (Linear) Search | Binary Search |
| :--- | :--- | :--- |
| **Data Requirement** | Can be unsorted or sorted | **Must be sorted** |
| **Worst-Case Time** | $\mathcal{O}(n)$ | $\mathcal{O}(\log n)$ |
| **Best-Case Time** | $\mathcal{O}(1)$ (Target is the first item) | $\mathcal{O}(1)$ (Target is the exact middle item) |
| **Space Complexity** | $\mathcal{O}(1)$ | $\mathcal{O}(1)$ (Iterative) / $\mathcal{O}(\log n)$ (Recursive) |
| **Scale Performance** | Performance degrades linearly with large datasets | Highly scalable (Checks 1 million items in max 20 steps) |

---

## 5. Architectural Guide: When to Use vs. Avoid

### ✅ When to Use
* **Static, Large Datasets:** When you have massive collections of data that do not change frequently, allowing you to pay the upfront cost of sorting once.

* **Frequent Lookups:** Exceptional for registries, libraries, systems, or dictionary databases that experience heavy read operations but minimal write operations.

* **Memory-Constrained Environments:** The iterative form provides constant memory overhead ($\mathcal{O}(1)$) making it safe for low-level systems.

* **Monotonic Functions:** Used in optimization problems ("Binary Search on Answer") where you look for the first valid threshold value across a valid numerical range.

### ❌ When to Avoid
* **Unsorted, Dynamic Data:** If data is constantly inserted or modified, the cost of continually sorting the array ($\mathcal{O}(n \log n)$) heavily eclipses any time saved by searching logarithmically.

* **Small Collections:** For tiny arrays (e.g., fewer than 10-20 elements), the mechanical setup overhead of binary search is slower than just running a simple linear scan.

* **Linked Lists:** Binary search relies heavily on "random access" ($\mathcal{O}(1)$ time to check any index pointer). Linked lists require sequential traversing to find the middle element, destroying the algorithmic benefit.

### 1. Binary Search 

```python
# binary search algorithm
# time complexity: O(log n)
# space complexity: O(1)

def binary_search(nums,target):
    left , right = 0, len(nums)-1
    
    while left <= right:
        mid = left + (right-left)//2
        
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            left = mid+1
        else:
            right = mid-1
    return -1
```

```python
arr1 =[-1,0,3,5,9,12]
tar1 = 12
tar2 = -2
print(f"Is {tar1} in the {arr1} : ",binary_search(arr1, tar1))
print(f"Is {tar2} in the {arr1} : ",binary_search(arr1, tar2))

```

### 2. Binary Search on Condition 

```python

def binary_search_condition(arr):
    left, right = 0, len(arr)-1
    
    while left < right:
        mid = left + (right-left)//2
        if arr[mid]:
            right = mid
        else:
            left = mid+1
    return left
```

```python
b_arr = [True, False, False, False, False, True, True, True]
print(f"First index of True in :-  {b_arr} : ",binary_search_condition(b_arr))
```
