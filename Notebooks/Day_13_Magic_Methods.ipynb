## Magic Method

- Magic methods in Python, also know as dunder methods (double underscores), are special methods that have a double underscore prefix and suffix. 
They are used to define behavior for built-in operations or functions.

- They are predefined Methods in Python that we can override to change the behavior of our objects. 
- Magic methods are defined using two underscores before and after the method name, such as `__init__`, `__str__`, `__add__`, `__eq__`, etc.



some magic methods in python:
- `__init__` : This is a special method that is called when an object is created. It initializes the attributes of the object.
- `__str__` : This is a special method that returns a string representation of the object. It is used to print the object.
- `__add__` : This is a special method that adds two objects together. It is used to perform addition on objects.
- `__eq__ `: This is a special method that checks if two objects are equal. It is used to compare objects.
- `__lt__` : This is a special method that checks if one object is less than another. It is used to compare objects.
- `__gt__` : This is a special method that checks if one object is greater than another. It is used to compare objects.
- `__repr__` : This is a special method that returns a string representation of the object. It is used to print the object.
- `__len__` : This is a special method that returns the length of the object. It is used to get the size of an object.
- `__getitem__` : This is a special method that allows you to access elements in an object using indexing or slicing.
- `__setitem__` : This is a special method that allows you to modify elements in an object using indexing or slicing.

```python
# Example 1:

class Car:
    def __init__(self, name, engine, year):
        self.name = name 
        self.engine = engine
        self.year = year
    
    def __str__(self):
        return f" Name: {self.name} | Engine: {self.engine} |  Year: {self.year}"
    
    def __repr__(self):
        return f"Car('{self.name}', '{self.engine}', {self.year})"
c = Car("Toyota", "V8", 2021)
print(c) # This will print the object in a readable format
print(repr(c)) # This will print the object in a more detailed format
```

```python
# Example 2: Custom Magic Methods 

class Counter:
    
    def __init__(self):
        self.value = 1
        
    def count_up(self):
        self.value +=1
        
    def count_down(self):
        if self.value > 0:
            self.value -= 1
            
    def __str__(self):
        return f"Count= {self.value} "
    
    def __add__(self, other):
        
        if isinstance(other, Counter):
            return self.value + other.value
        if isinstance(other, int):
            return self.value + other
        return NotImplemented

count1 = Counter()
count2 = Counter()

count1.count_up()
count2.count_down()

print("Count 1 : ",count1.value) # Output: 2
print("Count 2 : ",count2.value) # Output: 0



print('Sum : ',count1 + count2) # By implementing __add__ method this + sign is valid to use 
print("Sum : ",count1 + 2)
```

```python
# __str__ vs __reps__

class Car:
    
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
        
    # This method is meant for a more detailed, unambiguous output
    def __repr__(self):    #(MOSTLY USED FOR DEBUGGING)
        return f"Car(Brand = '{self.brand}', Model = '{self.model}', Year='{self.year}')"  
    
    # This method is meant for user-friendly output 
    def __str__(self):
        return f"{self.year} | {self.model} | {self.brand}"


my_car = Car('Porch', 'V8 Turbo', 2026)
print(my_car) # Always looks for __str__ method if not there it looks for __repr__
print(repr(my_car))
```

```python
# Arithmetic Operations 

class InventoryItems:
    
    """ This class demonstrate operator overloading for Inventory Management """
    
    def __init__(self, name, quantity):
        self.name = name
        self.quantity = quantity
        
    def __repr__(self):
            return f"InventoryItem(name = {self.name}, quantity = {self.quantity})"
    
    # Arithmetic Addition
    def __add__(self, other):
        if isinstance(other, InventoryItems) and self.name == other.name :
            return InventoryItems(self.name , self.quantity + other.quantity)
        raise ValueError("can't add items of different types !!")
    
    # Arithmetic Subtraction
    def __sub__(self, other):
        if isinstance(other, InventoryItems) and self.name == other.name:
            if self.quantity >= other.quantity:
                return InventoryItems(self.name , self.quantity - other.quantity)
            raise ValueError("Can't subtract more than the available quantity")
        raise ValueError ("Can't Subtract items of different types")
    
    # Arithmetic Multiplication
    def __mul__(self, factor):
        if isinstance(factor, (int, float)):
            return InventoryItems(self.name , int(self.quantity * factor))
        raise ValueError ("Multiplication factor must be a number")
    
    # Arithmetic Division
    def __truediv__(self, factor):
        if isinstance(factor, (int, float)) and factor !=0:
            return InventoryItems(self.name, int(self.quantity / factor))
        
        raise ValueError("ZeroDivisionError")
    
    
    # Comparison Operator
    
    def __eq__(self, other):
        if isinstance(other, InventoryItems):
            return self.name == other.name and self.quantity == other.quantity
        return False
    
    def __lt__(self, other):
        if isinstance(other, InventoryItems) and self.name == other.name:
            return self.quantity < other.quantity
        raise ValueError("Can't Compare items of different types!")
    
    def __gt__(self, other):
        if isinstance(other, InventoryItems) and self.name == other.name:
            return self.quantity > other.quantity
        raise ValueError ("Can't compare items of types")
    

# Creating some inventory items 

item1 = InventoryItems('Apple', 20)
item2 = InventoryItems('Orange', 1)
item3 = InventoryItems('Apple', 10)

# Adding Quantities of the same item
result1 = item1 + item3
print(result1)

# Subtracting Quantities of the same item 
result2 = item1 - item3
print(result2)

# Multiply item quantity by a given factor
result_mul = item1 * 2 
print(result_mul)

# Dividing item quantity by a given number 
result_div = item3 / 2
print(result_div)


# Let's try to add items of different types
print()
print("Adding Items of Different Types! ")
try : 
    result_invalid = item1 + item2
except ValueError as e:
    print(e)    


# Comparison operator
print(item1 > item3 )

try:
    print(item1 > item2)
except ValueError as e:
    print(e)    
```

```python
# Implementation of Linked List

class Node:
    
    def __init__(self, value):
        self.value = value
        self.next = None

class LinkedList:
    
    def __init__(self):
        self.head = None
        self.size = 0
    
    def __len__(self):
        """ Defines the Behavior of len() in list """
        return self.size
    
    def __getitem__(self, key):
        """ Enables indexing (obj[key]) """
        
        if key < 0 or key >= self.size:
            raise IndexError("Index Out Of Bound")
        current = self.head
        
        for _ in range(key):
            current = current.next
        return current.value    
    
    def __setitem__(self, key, value):
        """ Enables items Assignment (obj[key] = value)"""
        
        if key < 0 or key >= self.size:
            raise IndexError("Index Out of Bound")
        
        current = self.head
        for _ in range(key):
            current = current.next
        current.value = value   
        
    def __delitem__(self, key):
        """ Deleting an item (del obj[index])"""
        
        if key < 0 or key >= self.size:
            raise IndexError("Index Out of range")
        
        if key == 0:
            self.head = self.head.next
        else:
            current = self.head 
            for _ in range(key - 1):
                current = current.next
            current.next = current.next.next
        self.size -= 1
        
    def __contains__(self, value):
        """Defines behavior for 'in' keyword """
        
        current = self.head
        while current:
            if current.value == value:
                return True
            current = current.next
            
        return False
    
    def append(self, value):
        """Add a new node to the end of the list"""
        new_node = Node(value)
        if not self.head:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node
        self.size +=1
        
    def __str__(self):
        """ User Defined String Representation"""
        
        values = []
        current = self.head
        
        while current:
            values.append(str(current.value))
            current = current.next
        return "-->".join(values)


l1 = LinkedList()
l1.append(10)
l1.append(20)
l1.append(30)

print(len(l1))
print(l1[0]," | ", l1[1], " | ", l1[2])

# Append Values
l1[1] = 25
print(l1[1])

# delete
del l1[1]
print(l1)

# check value
print(30 in l1)
```
