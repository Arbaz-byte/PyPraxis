```python
# Array of Edges (Directed) [start, end]

n = 8 # Number of vertices
A = [[0,1], [1,2], [0,3], [3,4], [3,6], [3,7], [4,2], [4,5], [5,2]] 

print(A)
```

```python
# Convert Array of Edges -> Adjacency Matrix

M = []
for i in range(n):
    M.append([0]*n)

display("Empty List of Edges: ",M)
for edge in A:
    start = edge[0]
    end = edge[1]
    M[start][end] = 1
    
    # Uncomment the below line to print the edges
    #print(start, " -> ", end)
    
    # Uncomment the below line to make the graph undirected
    #M[end][start] = 1

display("Adjacency Matrix: ",M)
```

```python
# Convert Array of Edges -> Adjacency List

from collections import defaultdict

L = defaultdict(list)
for edge in A:
    start = edge[0]
    end = edge[1]
    L[start].append(end)
    
    # Uncomment the below line to print the edges
    #print(start, " -> ", end)
    
    # Uncomment the below line to make the graph undirected
    #L[end].append(start)

display(L)
# Easy to find connected components, shortest path etc.
print("Node 3 is connected to : ",L[3])
```

```python
# DFS with Recursion 
# Time Complexity: O(V + E) where V is number of vertices and E is number of edges
# Space Complexity: O(V) where V is number of vertices

def dfs_recursion(node):
    print(node)
    for neighbor in L[node]:
        if neighbor not in seen:
            seen.add(neighbor)
            dfs_recursion(neighbor)


source = 0
seen = set()
seen.add(source)

dfs_recursion(source)
```

```python
# Iterative DFS with stack 
# Time Complexity: O(V + E) where V is number of vertices and E is number of edges
# Space Complexity: O(V) where V is number of vertices

source = 0
seen = set()
seen.add(source)

stack = [source]

while stack:
    node = stack.pop()
    print(node)
    
    for neighbor in L[node]:
        if neighbor not in seen:
            seen.add(neighbor)
            stack.append(neighbor)
        
```

```python
# BFS (Queue) with queue
# Time Complexity: O(V + E) where V is number of vertices and E is number of edges
# Space Complexity: O(V) where V is number of vertices

from collections import deque

source=0
seen = set()
seen.add(source)

q = deque()
q.append(source)

while q :
    node = q.popleft()
    print(node)
    
    for neighbor in L[node]:
        if neighbor not in seen:
            seen.add(neighbor)
            q.append(neighbor)
```

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.neighbors = []
        
    def __str__(self):
        return f'Node({self.val})'
    
    def display(self):
        connections = [node.val for node in self.neighbors]
        return f'{self.val} connected to {connections}'
    

A = Node(1)
B = Node(2)
C = Node(3)
D = Node(4)

A.neighbors.append(B)
B.neighbors.append(A)
C.neighbors.append(D)

print(A.display())
print(B.display())
print(C.display())
```
