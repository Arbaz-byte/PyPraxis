# Encapsulation And Abstraction

- Encapsulation and abstraction are two fundamental concepts in object-oriented programming (OOP) that help in managing complexity and enhancing code maintainability.

- **Encapsulation** refers to the bundling of data (attributes) and methods (functions) that operate on that data into a single unit, typically a class. It restricts direct access to some of an object's components, which can prevent the accidental modification of data. This is often achieved through access modifiers like private, protected, and public.

- **Abstraction** is the process of hiding the complex implementation details and showing only the essential features of an object. It allows programmers to focus on what an object does instead of how it does it. Abstraction can be achieved through abstract classes and interfaces.



## 1. Encapsulation 

Encapsulation is the concept of wrapping data(variables) and methods(functions) into a single unit called class. It restricts direct access to some of an object's components, which can prevent the accidental modification of data.

- **Public** - Accessible from anywhere
- **Protected** - Accessible within the class and its subclasses
- **Private** - Accessible only within the class

Access modifiers are used to define the visibility of class members (attributes and methods). In Python, we use naming conventions to indicate the intended access level:
- **Public**: No underscore prefix (e.g., `attribute`)
- **Protected**: Single underscore prefix (e.g., `_attribute`)
- **Private**: Double underscore prefix (e.g., `__attribute`)


````python
# Encapsulation with Public, Private and Protected variable

class Person:
    def __init__(self, name, age, blood_group, credit_score):
        
        self.name = name # Public variable
        self.age = age # public variable
        
        self.__blood_group = blood_group # Private variable
        self._credit_score = credit_score # Protected variable

# Function defining Polymorphism
def get_name(Person):
    return Person.name

# Accessing Protected variable from derived class
class Employee(Person):
    def __init__(self, name, age, blood_group, credit_score):
        super().__init__(name, age, blood_group, credit_score)
        


# ============= Objects ====================================
p1 = Person("khan", 30, 'O-', 840)
employee = Employee("Arbaz", 30, 'AB+', 900)


# ============= Accessing Variables =========================

print(get_name(p1))         # Accessing public method directly
print(p1.age)              # Accessing public attribute directly
try:
    print(p1.blood_group) # Trying to access private attribute directly
except AttributeError as e:
    print(e)    

try:
    print(p1.credit_score) # Trying to access protected attribute directly
except AttributeError as e:
    print(e)

# ============= Accessing private and protected variables =========================

# Accessing Private attribute (not a good practice)
print(p1._Person__blood_group)
            
# Protected variable can be accessed but not recommended
print(p1._credit_score)

# Accessing inherited method from derived class
print(employee._credit_score)  # Accessing protected variable from derived class
````

````python
# Encapsulation with getter and setter methods

class Person:
    
    # Constructor 
    def __init__(self, name, age, id):
        self.__name = name # private access modifier or variable
        self.__age = age   # private access modifier or variable
        self.__id = id     # private access modifier or variable
    
    # Getter method to get the value of private attribute
    def get_info(self):
        return f"Name: {self.__name}, Age: {self.__age}, ID: {self.__id}"
    
    # Setter method to set the value of private attribute
    def set_info(self, name, age, id):
        self.__name = name
        if age > 0:
            self.__age = age
        else:
            print("Age must be greater than zero")
        self.__id = id

# Creating an instance of the Person class
person1 = Person("peter", 30, "EEI2345J")

# Accessing the info using getter method
print(person1.get_info())

# Modifying the info using setter method
person1.set_info("Peter Parker", 30, "EEI2345K")
print("After Modification :: ",person1.get_info())
````

## 2. Abstraction
Abstraction is the concept of hiding the complex implementation details and showing only the essential features of an object. It allows programmers to focus on what an object does instead of how it does it.

In Python, we can use the `abc` module to define abstract classes and methods. An abstract class is a class that cannot be instantiated and is meant to be subclassed. It can contain abstract methods, which are methods that are declared but contain no implementation. Subclasses must provide implementations for all abstract methods.

Here's an example of how to use the `abc` module to define an abstract class and method:
```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def make_sound(self):
        pass
``` 

````python
from abc import ABC, abstractmethod

## Abstract Base Class 
class Vehicle(ABC):
    
    def drive(self):
        print("Driving Mode Activated!!")
    
    @abstractmethod
    def start_engine(self):
        pass

# Subclasses of the Abstract Base Class
class Car(Vehicle):
    def start_engine(self):
        print("Engine Started! ")

# Method Defining Polymorphism

def operate_vehicle(Vehicle):
    return Vehicle.start_engine()

car1 = Car()
operate_vehicle(car1) 
````
