## Sets


```python
# to create a set 
a = {1,2,3}
print(type(a))
```

```python
empty_set = set()
print(type(empty_set))
```

```python
typecast_set = set([1,2,2])
print(type(empty_set))
print(typecast_set) # ignore duplicate and unordered
```

```python
# Set operation

# Add elements
examp_set = {1,2,3,4}
examp_set.add(5)
print(examp_set)

# Remove element
examp_set.remove(1)
print(examp_set)

# Removing element that didn't exist 
examp_set.discard(100)
print(examp_set) # If element is not found will print the same set

# Removing random element 
examp_set.pop()
print(examp_set) # follows the rule of first in first out 

# to clear the set 
examp_set.clear()
print(examp_set) # will return empty set
```

```python
# Set membership set
new_set = {1,2,3,4,5,6}

print(3 in new_set)# return boolean
print(10 in new_set)
```

```python
# Mathematical operations on set 

set1= set([i for i in range(1,11)])
set2 = set([i for i in range(3,-11,-1)])
set3 = set([i for i in range(11,21)])


## Union
union_set = set1.union(set3)
print(union_set)

## Intersection
intersection_set = set1.intersection(set2)
print(intersection_set)

## Intersection update
set1.intersection_update(set2)
print(set1) #update set1 with common elements in set1 and set2

## Difference of set
diff_set = union_set.difference(set2)
print(diff_set) # elements which are not present in first but in set2 (removes common elements)

## Symmetric Diff 
symm_diff_set = union_set.symmetric_difference(set3)
print(symm_diff_set)# Unique elements from both the sets common elements are removed



```

```python
# Sets Methods 

set1 = {1,2,3,4,5,6,7}
set2= {2,4,6}
set3 = {20,40,60}
## Check for subset
print("Is set1 is subset of set2 : ",set1.issubset(set2))
print("Is set2 is subset of set2 : ",set2.issubset(set1))

# Check for superset
print(set1.issuperset(set2))

# Check for disjoint set
print(set1.isdisjoint(set3))# no common elements
print(set1.isdisjoint(set2))# common elements
```

```python
### counting the unique elements in the text

txt = "The unique elements are only allowed , but the order is not!"
words = txt.split()

# convert the list of words into set for unique elements 
unique_words = set(words)
print(unique_words)
print(len(unique_words))
```

## Dictionaries 

- It is unordered collection of items. They store data in key-value pairs.
- keys must be unique and immutable(e.g. string, numbers, or tuple), while values can be of any type

```python
# Creating dictionaries

# 1. 
empty_dict = {}
print(type(empty_dict))

# 2. 
empty_dict1 = dict()
print(type(empty_dict1))


```

```python
# Accessing Dictionary elements 
student = {'name' : "Khan", "Age" : 24 , "gender": "M", "Grade" : 'A+'}
print("name of student : ", student['name'])
print("grade of student : ", student['Grade'])


id = {'name': "Arbaz", 'Age': 24 , 'name': 90}
print(id) # takes the current values of name and allow only unique keys 


# Accessing using get() method
print("Age : ",student.get('Age'))
print("University : ",student.get('university', "Not Available!")) # if key is not present


```

```python
# Modifying Dictionary Elements as it is mutable
student1 = {'name' : "Arbaz Khan", "Age" : 24 , "gender": "M", "Grade" : 'A+'}

## Add new key
student1["Blood group"]="A+"
student1['lang'] = "English"
print(student1)

## Modify value 
student1['Age']=23
print(student1)

# Delete key
del student1['lang'] # delete the lang key-value pair
print(student1)
```

```python
# Dictionary Methods
student2 = {'name' : "Mark", "Age" : 22 , "gender": "M", "Grade" : 'B+'}

# Get keys and  value
keys = student2.keys()
values = student2.values()
print("keys :",keys)
print("values : ",values)

# Get key-value pairs 
items = student2.items()
print("key-value pairs :",items) # List of tuples 
```

```python

# Problem in copy dictionary
student2 = {'name' : "Mark", "Age" : 22 , "gender": "M", "Grade" : 'B+'}

student_copy = student2 # both are pointing the same memory location 
student2["name"]= "Mark 2"

print(student2)
print(student_copy) # so it change both of them 


# shallow copy
student_copy1 = student2.copy() # points to different memory location
student2["name"]= "Mark 3"

print(student2)
print(student_copy1) # so it change for student2 only
```

```python
# Iterating over Dictionary

SuperHero = {'name' : "Thor", "Age" : 1500 , "gender": "M", "Grade" : 'No Need!!!'}


# 1. For loop (dict --> keys, values, and items)

# a.) key
for keys in SuperHero.keys():
    print(keys)


print("--"*20)


# b.) Values
for value in SuperHero.values():
    print(value)


print("--"*20)

# c.) Items 
for keys,value in SuperHero.items():
    print(f"{keys}: {value} ")

```

```python
# Nested Dictionary 

Marvel ={
    "SuperHero1" :{'name' : "Thor", "Age" : 1500 , "gender": "M", "Grade" : 'No Need!!!'},
    "SuperHero2" :{'name' : "Iron Man", "Age" : 45 , "gender": "M", "Grade" : 'Seriously ?'},
    "SuperHero3" :{'name' : "Bruce", "Age" : 55, "gender": "M", "Grade" : 'A++'}
}

## Access nested dictionary
print(Marvel['SuperHero2']['name'])
print(Marvel['SuperHero2']['Grade'])

# key-value 
print(Marvel.items())
```

```python
# Iterating to Nested dictionary

for marvel_id, marvel_info in Marvel.items():
    print(f"{marvel_id} : {marvel_info}")
    
    for key,value in marvel_info.items():
        print(f"{key} : {value}")
    print("-----"*20)    
    
```

```python
# Dictionary Comprehension

squares = {i:i**2 for i in range(10)}
print(squares)

# Conditional Dictionary Comprehension
Square_even = {i:i**2 for i in range(1,11) if i%2==0}
print("Even Square : ", Square_even)

```

```python
# Q. Use a dictionary to count the frequency of elements in given list 

numbers = [1,2,2,3,3,3,3,4,4,5,5,5,5,5,6,6,7,7]
frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num]+=1 # Count frequency of key
    else:
        frequency[num]=1 # Make key
        
print(frequency)            
```

```python
# Marge dictionaries 

dict1 = {'a': 1 , 'b' : 2}
dict2 = {'c': 3, 'd' : 4}

merged_dict = {**dict1, **dict2} # ** --> Takes key value pairs 
print(merged_dict)
```
