## 1. Introduction to Python OOPs

Python is a high-level, interpreted programming language that emphasizes code readability and simplicity. It supports object-oriented programming (OOP) as one of its core features.

- OOPs stand for Object-Oriented Programming (OOP)
- OOPs is a programming paradigm that uses objects and classes to structure programs and provides a way to structure data and behavior.





### key Features of Python OOPs:

- **Classes and Objects**: Classes are blueprints for creating objects, which are instances of a class. A class defines the properties (attributes) and behaviors (methods) that its objects will have. In Python, you can define a class using the `class` keyword, and create objects by instantiating the class.

- **Methods**: Methods are functions defined within a class. They can access the attributes of the class and perform operations on them. In Python, methods are defined using the `def` keyword, and the first parameter of a method is typically `self`, which refers to the instance of the class.

- **Attributes**: Attributes are variables defined within a class. They can hold data specific to the object and can be accessed and modified using methods. In Python, attributes can be defined within the `__init__` method, which is a special method that is called when an object is created.

- **Polymorphism**: Polymorphism is the ability of an object to take on many forms. It allows different objects to be treated as instances of the same class through a common interface. In Python, polymorphism can be achieved through method overriding and operator overloading. 

- **Encapsulation**: Encapsulation is the process of bundling data and methods that operate on that data within a single unit called a class. It helps in hiding the internal details of an object and provides a public interface for interacting with it. In Python, encapsulation can be achieved using access modifiers such as `public`, `private`, and `protected`.   

- **Inheritance**: Inheritance is a mechanism that allows a class to inherit properties and methods from another class. This promotes code reusability and allows you to create new classes based on existing ones. In Python, inheritance is achieved by defining a subclass that inherits from a superclass.  

- **Polymorphism**: Polymorphism is the ability of an object to take on many forms. It allows different objects to be treated as instances of the same class through a common interface. In Python, polymorphism can be achieved through method overriding and operator overloading.

- **Abstraction**: Abstraction is the process of hiding unnecessary details and showing only essential information to the user. In Python, abstraction can be achieved through the use of abstract classes and methods. This allows you to create a simplified interface for complex systems, making it easier for other developers to use your code.

- **Data Hiding**: Data hiding is the practice of restricting access to an object's internal state and behavior. It helps in preventing unauthorized access and modification of an object's data. In Python, data hiding can be achieved using access modifiers such as `private` and `protected`.  

- **Encapsulation**: Encapsulation is the process of bundling related data and functions into a single unit called a class. This allows you to create objects that have their own state and behavior, making it easier to manage and manipulate data in your code. In Python, encapsulation can be achieved using classes and objects.

# `__init__()` Method

All classes have a built-in function called `__init__()`, which is always executed when a new object is created from a class. This function can be used to assign initial values to the attributes of the object. It always executed when class is instantiated. 
 It takes at least one argument, which is the name of the object being created. 




```python
### Instance Variable 

class dog:
    
    ##Constructor
    def __init__(self, name, age):
        
        # self.__ --> Instance variable (Properties of class dog)
        self.name = name
        self.age = age

## Create object 
dog1 = dog("kripto","100")
print("Memory Location : ",dog1) # Retrieving  Attributes 
print("Name : ", dog1.name , " | Age : ",dog1.age)

```

```python
### Instance Methods

class cat:
    
    ## Constructor 
    def __init__(self, name, age):
        self.name = name # (properties of cat class share with all)
        self.age = age
    
    
    ## Instance Methods
    def move(self):  # Method shared by all the cat objects 
        print(f"{self.name} is Moving!")



persian_cat = cat("Simba", '6')   
print("Name :",persian_cat.name," | Age :" ,persian_cat.age )
persian_cat.move()

indian_cat = cat("meow", '4')
indian_cat.move()
```

```python
# Example 1: Bank Account

class BankAccount:
    
    def __init__(self, name, balance = 0):
        self.name=name
        self.balance = balance
    
    ## Instance Method for deposit
    def deposit(self, amount):
        self.balance += amount 
        print(f"{amount} is deposited. New Balance is {self.balance}")
        
    ## Instance Method for withdraw 
    def withdraw( self, amount):
        if amount>self.balance:
            print("Insufficient Funds! ")
        else:
            self.balance -= amount
            print(f"{amount} is withdrawn. Remaining balance is {self.balance}")
    
    ## Instance Method for balance enquiry 
    def check_balance(self): 
        return self.balance

## Create an Object

acc1 = BankAccount("Tony", 1000)
print(f"Account Balance of {acc1.name} is : ",acc1.check_balance())
print()


## call instance Method
acc1.deposit(10)
print()

# Withdraw 
acc1.withdraw(10000)
print()

acc1.withdraw(1000)
```
