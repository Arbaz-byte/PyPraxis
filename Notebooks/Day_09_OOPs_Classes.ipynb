# Python OOPs

## 1. Introduction to Python OOPs

Python is a high-level, interpreted programming language that emphasizes code readability and simplicity. It supports object-oriented programming (OOP) as one of its core features.

- OOPs stand for Object-Oriented Programming (OOP)
- OOPs is a programming paradigm that uses objects and classes to structure programs and provides a way to structure data and behavior.


### Use Cases for Python OOPs:



- **Code Reusability**: OOPs allows you to create classes and objects that can be reused across different parts of your codebase. This reduces redundancy and makes it easier to maintain and update your code.

- **Modularity**: OOPs helps in organizing code into smaller, manageable modules. Each module can contain related functions, classes, and variables. This makes it easier to manage large projects with multiple developers working on different parts of the project.

- **Encapsulation**: Encapsulation is the practice of bundling related data and functions within a single unit called a class. It helps in hiding the implementation details of a class, making it easier for other developers to understand and use the class.

Also it help to keep your code clean, organized, and maintainable.(DRY - Don't Repeat Yourself)

- **Polymorphism**: Polymorphism is the ability of an object to take on many forms. In Python, polymorphism allows you to define methods in a class that can be overridden by subclasses. This enables you to create more flexible and reusable code.

- **Abstraction**: Abstraction is the process of hiding unnecessary details and showing only essential information to the user. In Python, abstraction can be achieved through the use of abstract classes and methods. This allows you to create a simplified interface for complex systems, making it easier for other developers to use your code.

- **Inheritance**: Inheritance is a mechanism that allows a class to inherit properties and methods from another class. This promotes code reusability and allows you to create new classes based on existing ones. In Python, inheritance is achieved by defining a subclass that inherits from a superclass.

- **Interfaces**: Interfaces are used to define a set of methods that a class must implement. Interfaces are used to ensure that classes adhere to certain rules and behaviors, making it easier for other developers to use your code. In Python, interfaces can be implemented using abstract base classes (ABCs) and the `abc` module.

- **Attributes**: Attributes are variables defined within a class. They can hold data specific to the object and can be accessed and modified using methods. In Python, attributes can be defined within the `__init__` method, which is a special method that is called when an object is created.

- **Polymorphism**: Polymorphism is the ability of an object to take on many forms. It allows different objects to be treated as instances of the same class through a common interface. In Python, polymorphism can be achieved through method overriding and operator overloading. 

- **Encapsulation**: Encapsulation is the process of bundling data and methods that operate on that data within a single unit called a class. It helps in hiding the internal details of an object and provides a public interface for interacting with it. In Python, encapsulation can be achieved using access modifiers such as `public`, `private`, and `protected`.   

- **Inheritance**: Inheritance is a mechanism that allows a class to inherit properties and methods from another class. ## What are Methods ?
Methods are functions defined inside a class. They are used to perform operations on the data stored in an object. 
They can access and modify the data of the object they belong to. 
Methods are called on objects using dot notation (object_name.method_name) 

## What is Constructor ?
A constructor is a special method that is used to initialize the attributes of an object when it is created.
It is called automatically when an object is created from a class. It can be defined using the `__init__` method. 



```python
class car: # class
    x=5 # property
    
c1= car() # Object of class car
print(c1) 
print(c1.x)   

# can also delete objects
del c1
try:
    print(c1)
except Exception as e:
    print(e)    

# Class with pass statement
class bike:
    pass
```
