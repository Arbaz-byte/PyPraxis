# File Operation - Read and Write 

- Python provided built-in functions and methods to read and write files, both text and binary.

```python
# Reading File 
def open_file(path, mode='r'):
    with open(path, mode) as file:  
        content = file.read()
        print(content)    
        
open_file(path='./package/test/test.txt')
```

```python
# Append File (Without Overwriting Mode)
with open('./package/test/test.txt', 'a') as file: # a -->append mode
    file.write('\nThis is a new line add using append mode')
    
# Reading File Again
open_file(path='./package/test/test.txt')
```

```python
# Writing File (Overwrite Mode)
with open('./package/test/test.txt', 'w') as file:
    file.write('The given file content has been changed!')  
    file.write(
        '\nThis method overwrite everything'
    )

# Reading File Again
open_file(path='./package/test/test.txt')
```

```python
# Writing and then reading a file
with open('./package/test/test1.txt', 'w+') as file:
    file.write('This is another example of writing and then reading a file')
    file.write('\n use w+ for this purpose')
    
    # Move the file cursor to the beginning of the file
    file.seek(0)
    
    # Read the content of the file
    content = file.read()
    print(content)
```

```python
# Writing a list of lines to a file

lines = ['\nLine 1\n', 'Line 2\n', 'Line 3\n']
with open('./package/test/test.txt', 'a') as file:
    file.writelines(lines)
    
# Reading File Again
open_file(path='./package/test/test.txt') 
```

```python
# Binary Files

data = b'\x00\x01\x02\x03\x04\x05\x06'
with open('./package/test/test.bin', 'wb') as file: # wb --> write binary mode
    file.write(data)


# Reading File Again
open_file(path='./package/test/test.bin', mode='rb')   

```

### Example 1: Reading a Text File and counting lines, words and characters
 


```python
def count_text_file(path):
    line_count = 0
    words_count = 0
    char_count = 0
    
    with open(path, 'r') as file:
        lines = file.readlines()  # Read all lines at once
        for line in lines:
            line_count += 1
            words_count += len(line.split())  # Count words in the current line
            char_count += sum(len(word) for word in line.split())  # Count characters in the current line
    
    return line_count, words_count, char_count

# Example usage
file_path = './package/test/test.txt'
line_count, words_count, char_count = count_text_file(file_path)

print(f" Lines : {line_count} |  Words : {words_count} | Characters : {char_count}")
```

```python
# Joining paths

dir_name = './package/Utilities'
file_name = 'test.txt'

# Creating Directory if it does not exist
if not os.path.exists(dir_name):
    os.mkdir(dir_name)

# Join the directory and file name to get the full path of the file
file_path = os.path.join(dir_name, file_name)


# Writing some content to the file
with open(file_path, 'w') as file:
    file.write('This is an example for file joining')

print(f"File Path: {file_path}")    

open_file(path=file_path)
```

```python
# Check if the file exists

path = './package/test/test.txt'

if os.path.exists(path):
    print(f"File exist at path : {path}")
else:
    print(f"File path does not exist!")    
```

```python
# checking if a path is a File or Director

path = './package/Utilities'

if os.path.isfile(path):
    print(f"The Path {path} is a file!")
elif os.path.isdir(path):
    print(f"The path {path} is a Directory")
else:
    print(f"The path {path} does not exist")        
```
