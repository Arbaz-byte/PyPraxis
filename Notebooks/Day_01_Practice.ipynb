```python
# Printing the Prime number in the given range 

# Range
n=10
for num in range(1,n+1):
    if num>1 : # As greater than 1
        for i in range(2, num):# List of divisors to check if divisible or not
            if num % i == 0:
                #check given number as more than two factors 
                break
        else:
            print(num)       
```

## List


```python
# List - ordered and mutable 
lst = [2, 4, 6, 8, 10, 12, 14]

print("-- Slicing --")
#Accessing/slicing elements 
print(lst[:])# all elements
print(lst[1:]) # for index 1 to all
print(lst[-2:]) # from 12 to 14
print(lst[-1:-3]) # bad indexing as no items after 14
print(lst[::2]) # take elements at step size of 2
print(lst[::-1]) # take elements at 1 step from last i.e. reverse the list


print('-- Modify ---')
# Modify elements
lst[1:3]="modify"
print(lst) # add as single char replacing from 4 to 6


print("\n--List Method--")
char = ['a','b','c','d','e','a'] # allow duplicate values

# Append
char.append('f') # add to last
print(char)

# Insert 
char.insert(7,'g')# inserts element in given index 
print(char)

# pop
rem_ele = char.pop() # removes and return the last element
print("poped :",rem_ele)
print(char)

# index
print("Index:",char.index('b'))

# remove 
char.remove('a') # removes the first occurance of a
print(char)

# count
char.append('a')
print("Count:",char.count('a'))

# sort
char.sort()
print("sorted :",char)

# reverse
char.reverse()
print("Revere: ",char)

# clear
char.clear() # removes all elements 
print(char)
```

```python
# ========================================

# iterating with index
number = [1, 2, 3, 4, 5]

for indx,num in enumerate(number):
    print(f"Index:{indx} | Element :{num}") 

# ========================================

print("---"*20)
# List Comprehension
lit=[]
for i in range(10):
    lit.append(i**2)
print(lit)

print("using list comprehension:")
lis = [x**2 for x in range(10)] 
print(lis)

# =====================================
print("---"*20)
numb = [x for x in range(1,10) if x%2 == 0]
print(numb)   

# =======================================
print("---"*20)
# Nested list comprehension
lst1 = [1, 2, 3, 4]
lst2 = ['a', 'b', 'c', 'd']

pair_lst = [(i,j) for i in lst1  for j in lst2] # j runs ith times
print(pair_lst)

pair_lst = [(i,j) for i in lst1 if i==2 for j in lst2] # j runs ith times
print(pair_lst)

pair_lst = [(i,j) for i in lst1  for j in lst2 if j=='c'] # j runs ith times
print(pair_lst)

# =======================================
print("---"*20)

# List comprehension with function call
words = ["apple", "Linux", "Windows"]

length_word = [len(w) for w in words ]
print(length_word)
```
