#  Python Generators

A **generator in Python** is a special type of function that returns an **iterator object**, allowing us to loop through a sequence of values **one at a time on demand**. Unlike standard functions that compute an entire result and return it all at once, generators produce values dynamically, making them exceptionally **memory-efficient** when handling large datasets or infinite streams.

---

### 💡 Key Differences: Yield vs. Return

The core mechanic that transforms a standard function into a generator is the **`yield` keyword**. 

| Feature | Standard Function (`return`) | Generator Function (`yield`) |
| :--- | :--- | :--- |
| **Execution** | Runs to completion, then exits entirely. | **Pauses and resumes** execution state at the `yield` line. |
| **Memory Impact** | Allocates memory for the **entire result set** at once. | Allocates memory only for **one item at a time**. |
| **Output Type** | Returns a specific data structure (like a `list` or `string`). | Returns a **generator iterator object**. |

---

### 🚀 Why Should You Use Them?

*   **Massive Memory Savings:** If we need to read a 10 GB log file, a standard function reading all lines at once will crash your program with a `MemoryError`. A generator processes it line-by-line, using virtually zero memory.

*   **Lazy Evaluation:** Values are only computed when we explicitly ask for them. This saves CPU cycles if we decide to break out of a loop early.

*   **Infinite Sequences:** Since values aren't stored in memory, generators can easily represent infinite structures (like generating endless Fibonacci numbers) without freezing your system.


## 1. Generator Function

### Prime Number

```python
import sys
def prime_numbers(n):
    
    """Generator function that yields prime numbers up to n."""
    for num in range(1, n+1):
        if num>1:
            for i in range(2,num):
                if num % i == 0:
                    break
            else:
                yield num

# Using the generator function
prime = prime_numbers(1000000000000000)
print(f"Memory usage: {sys.getsizeof(prime)} bytes") # This clearly shows that no matter what the n is it takes constant memory!


for prime in prime_numbers(10):
    print(prime)
```

### Square of n Numbers

```python
# Using the generator function with a loop
def square_num_gen(n):
    """This Generator Function yields the squares of numbers up to n."""
    for num in range(1, n+1):
        yield num**2

square = square_num_gen(5)
next(square)
next(square)
next(square)

for i in square: # The output will be 16,25 because the generator function is not executed until it is called. 
    print(i)
```

### Custom range vs Generator Function

```python
class custom_range:
    
    def __init__(self, start, end, step=1):
        self.start = start
        self.end = end
        self.step = step
        
    def __iter__(self):
        return custom_range_iterator(self)

class custom_range_iterator:
    
    def __init__(self, iterable_obj):
        self.iterable = iterable_obj

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.iterable.start < self.iterable.end:
            current_value = self.iterable.start
            self.iterable.start += self.iterable.step
            return current_value
        else:
            raise StopIteration


for i in custom_range(1, 10):
    print(i)
```

```python
# Custom Range Generator Function
def custom_range_gen(start, end, step=1):
    
    for i in range(start, end, step):
        yield i

for i in custom_range_gen(1, 10):
    print(i)
```


## 2. Generator Expressions

```python
# List comprehension (Creates the whole list in memory immediately)
squares_list = [x**2 for x in range(1000000)] 

# Generator expression (Computes values on-the-fly as you loop)
squares_gen = (x**2 for x in range(1000000)) 

# Check the size of memory used by each list and generator
print("Size of squares_list:", sys.getsizeof(squares_list),"Bytes")
print("Size of squares_gen:", sys.getsizeof(squares_gen),"Bytes")
```

### Representing Infinite stream

```python
def infinite_sequence_even():
    n = 0
    while True:
        yield n
        n+=2
        
# Example usage of the infinite sequence function
generator = infinite_sequence_even()

for i in range(10):
    print(next(generator))
```

### Chaining Generators

```python
def fibonacci(n):
    x,y = 0,1
    
    for _ in range(n):
        x, y = y, x+y
        yield x

def square(n):
    for num in n:
        yield num**2


# Example usage of the chained generators
generator = fibonacci(5)
for i in generator:
    print(i)


# Sum of square of n fibonacci numbers
n=10
sum_of_squares = sum(square(fibonacci(n)))
print(f"Sum of squares of first {n} Fibonacci numbers: {sum_of_squares}")
```
