## What is an Iteration ?

Iteration in general refers to the process of repeating a set of instructions or operations. It involves iterating over a sequence, such as a list, tuple, or string to perform a specific action for each element.

```python
# Example: Iterating over a list

fruits = ["apple", "banana", "cherry"]

for fruit in fruits:
    print(fruit)
```

## What is Iterator ?

An iterator is an object that represents the current position of iteration over a sequence(don't store them in memory). It provides methods to access the elements of the sequence and move through them one by one.

```python
import sys
# Example : How iterator works


# With list comprehension
num = [x for x in range(1, 100000)] # Creating an iterator object containing numbers from 1 to 9999
print(sys.getsizeof(num)/(1024),"KB") # Size of the list in bytes

# with range function (generator)
num = range(1, 100000) # Creating an iterator
print(sys.getsizeof(num)/(1024),"KB")


# Clearly we can see that using a generator is more memory efficient than creating a list.
# This is because generators generate elements on-the-fly and do not store them in memory. 
# They are useful when dealing with large datasets or when you want to save memory space.
```

## What is Iterable ?

An iterable is an object that can be iterated over, such as a list, tuple or string. It provides methods to access the elements of the sequence and move through them one by one. It generates an iterator when passed to iter() method.

### Point to remember:

- Every iterable has an iterator.
- An iterator is an object that can be used to iterate over the elements of a sequence.

- Not all Iterables are iterators. For example, strings are not iterators because they do not have a next() method.
- The next() function returns the next item in the sequence. If there are no more items in the sequence, it raises a StopIteration exception.

- An iterator can be created from an iterable using the iter() function.
- The iter() function returns an iterator object that can be used to iterate over the elements of a sequence.

### NOTE :

- Every iterable has an **iter function**.
- Every iterator has a **next function**.



```python
list_num = [1, 2, 3, 4, 5] # Iterable object
iter_obj = iter(list_num) # Iterator object

print(type(list_num))
print(type(iter_obj)) 
print(next(iter_obj)) # Output: 1
print(next(iter_obj)) # Output: 2
```

## How python Loop Works 

```python
num = [1,2,3,4,5]

# Step 1:  Fetch the iterator object from the iterable using iter() function.
iter_obj = iter(num)

# Step 2: Check if the iterator is exhausted by calling next() function. If it raises StopIteration exception, then the loop terminates.
while True:
    try:
        # Step 3: Fetch the next element from the iterator using next() function.
        print(next(iter_obj))
    except StopIteration:
        break   

```

## Let's Make our own for loop

```python
def custom_loop(iterable):
    iter_obj = iter(iterable)
    while True:
        try:
            print(next(iter_obj))
        except StopIteration:
            break

a = [1,2,3]
b = range(1,5)
c = (2,4,6)
d = {0:1, 1:2}

custom_loop(a) # Output: 1 2 3 
custom_loop(b) # Output: 1 2 3 4 
custom_loop(c) # Output : 2 4 6 
custom_loop(d) # Output: 0 1 1
```

## Creating Own range() function

```python
class custom_range:
    
    def __init__(self, start, end, step=1):
        self.start = start
        self.end = end
        self.step = step
        
    def __iter__(self):
        return custom_range_iterator(self)

class custom_range_iterator:
    
    def __init__(self, iterable_obj):
        self.iterable = iterable_obj

    def __iter__(self):
        return self
    
    def __next__(self):
        if self.iterable.start < self.iterable.end:
            current_value = self.iterable.start
            self.iterable.start += self.iterable.step
            return current_value
        else:
            raise StopIteration

```

```python
# Creating an instance of custom_range class
custom_obj = custom_range(1,1000000, step=1)
print("Memory Usage : ",sys.getsizeof(custom_obj)/1024, "KB") # Memory Usage


for i in custom_range(1,5,step=1):
    print(i) 
```
