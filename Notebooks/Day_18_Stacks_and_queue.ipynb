# Stack Data Structure

A **Stack** is a linear data structure that follows the **LIFO (Last In, First Out)** principle. This means the last element added to the stack will be the first one to be removed. 

---

### Core Operations & Internal Mechanics

Natively, a stack can be built using either a **Static Array** or a **Dynamic Linked List**. Below is how the core operations work under the hood for both implementations.

#### 1. Push

Adds an element to the top of the stack.
* **Array Built:** It increments a pointer/index called `top` by 1 and inserts the element at `array[top]`. If `top` reaches the maximum capacity of the array, it triggers a **Stack Overflow** error.

* **Linked List Built:** It creates a new node. The `next` pointer of this new node points to the current `head` (the current top). Then, the `head` pointer is updated to point to this new node.

#### 2. Pop

Removes and returns the top element from the stack.
* **Array Built:** It checks if the stack is empty (`top == -1`). If it is, it triggers a **Stack Underflow** error. Otherwise, it retrieves the value at `array[top]` and decrements the `top` index by 1.

* **Linked List Built:** It checks if `head` is null (underflow). Otherwise, it saves the data from the `head` node, moves the `head` pointer to `head.next`, and deletes/frees the old top node.

#### 3. Peek / Top

Returns the value of the top element without removing it.
* **Array Built:** Simply returns `array[top]`.
* **Linked List Built:** Simply returns `head.data`.

#### 4. isEmpty

Checks if the stack has no elements.
* **Array Built:** Returns `true` if `top == -1`.
* **Linked List Built:** Returns `true` if `head == null`.

---

### Complexity Analysis

Here is the time and space complexity for stack operations. Because stacks only interact with the very end (the "top") of the data structure, their primary operations are incredibly fast.

| Operation | Time Complexity | Space Complexity |
| :--- | :--- | :--- |
| **Push** | O(1) | O(1) |
| **Pop** | O(1) | O(1) |
| **Peek / Top** | O(1) | O(1) |
| **isEmpty** | O(1) | O(1) |
| **Search / Access** | O(N) | O(1) |

*Note: Searching or accessing an element deep inside a stack requires popping elements one by one until you find it, which takes linear time O(N). Total space complexity for storing N elements is O(N).*

---

### Native Structural Comparison

Choosing how to build a stack natively changes how it handles memory layout and size limits:

| Feature | Array-Based Stack | Linked List-Based Stack |
| :--- | :--- | :--- |
| **Memory Allocation** | Contiguous memory allocation. | Non-contiguous memory (scattered). |
| **Size Flexibility** | Fixed size (unless using a dynamic array like Vector/ArrayList, which introduces occasional O(N) resizing costs). | Dynamic size (grows and shrinks seamlessly on demand). |
| **Memory Overhead** | Low overhead (only stores the raw data and one integer pointer). | Higher overhead (each element requires extra memory to store a pointer to the next node). |


## 1. Stack Implementation 

### 1.1 Array Based Stack (using Python's Native Dynamic List)

```python
class ArrayStack:
    
    def __init__(self):
        # Initialize an empty list to store the stack elements
        self.stack = []
        self.min = []
        self.max = []
    
    def __str__(self):
        return str(self.stack)
    
    def push(self, item):
        # Add an item to the top of the stack
        self.stack.append(item) # O(1) time complexity
        
        # Update min and max lists
        self.min.append(
            item if not self.min else min(item, self.min[-1])
        )
        self.max.append(
            item if not self.max else max(item, self.max[-1])
        )
    
    def pop(self):
        # Remove and return the item from the top of the stack. Raises Underflow error if empty.
        if self.is_empty():
            raise IndexError("Underflow: Cannot pop from an empty stack")
        
        popped = self.stack.pop()  # O(1) time complexity
        
        # Update min and max lists
        self.min.pop()
        self.max.pop()
        return popped  
    
    def minElement(self):
        # Return the minimum element in the stack. Raises Underflow error if empty.
        if self.is_empty():
            raise IndexError("Underflow: Cannot get minimum element from an empty stack")
        return self.min[-1]
    
    def maxElement(self):
        # Return the maximum element in the stack. Raises Underflow error if empty
        if self.is_empty():
            raise IndexError("Underflow: Cannot get maximum element from an empty stack")
        return self.max[-1]
    
    def peek(self):
        # Return the item at the top of the stack without removing it. Raises Underflow error if empty.
        if self.is_empty():
            raise IndexError("Underflow: Cannot peek from an empty stack")
        return self.stack[-1]

    def is_empty(self):
        # Check if the stack is empty
        return len(self.stack) == 0
    
    def size(self):
        # Return the current number of elements in the stack
        return len(self.stack)


# Example usage:
stack = ArrayStack()

for i in range(1, 5):
    stack.push(i**2)

print("Stack : ", stack)
print("Type : ", type(stack).__name__)

print("\nTop element:", stack.peek())
print("Size of stack:", stack.size())
print("Empty check: ", stack.is_empty())

print(stack.pop())
print("Top Element: ", stack.peek())
print(stack)
print("Min Element:", stack.minElement())
print("Max Element:", stack.maxElement())
```

### 1.2 Linked List-Based Stack

This implementation uses custom data nodes. It consumes memory dynamically on demand  and does not require contiguous memory allocation.

```python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedListStack:
    def __init__(self):
        # Initialize the stack with a dummy node
        self.head = Node
        self._size = 0
    
    def push(self, item):
        # Create a new node and set its data to 'item' and next pointer to the current head of the list
        new_node = Node(item)
        new_node.next = self.head
        
        # Move the top pointer to the new node
        self.head = new_node
        self._size += 1
    
    def pop(self):
        # Removes the top node and return its data
        if self.is_empty():
            raise IndexError("Stack Underflow : Can't pop from an empty stack")
        
        popped_data = self.head.next
        
        # Sever the reference to the old top node
        
        self.head = self.head.next
        self._size -= 1
        return popped_data
    
    def peek(self):
        # Look at the top node's data without altering the structure
        if self.is_empty():
            raise IndexError("Stack Underflow : Can't peek from an empty stack")
        return self.head.data
    
    def is_empty(self):
        # Check if the list is empty
        return self._size == 0
    
    def size(self):
        # Return the number of elements in the list
        return self._size


ll_stack = LinkedListStack()
ll_stack.push(1)
ll_stack.push(2)
print(ll_stack.pop())  
print(ll_stack.peek())  
print(ll_stack.is_empty())  
print(ll_stack.size())   
```

# Heap Data Structure

A **Heap** is a specialized tree-based data structure that satisfies the **Heap Property**. It is a **Complete Binary Tree**, meaning all levels are completely filled except possibly the last level, which is filled from left to right.

---

### Types of Heaps

1. **Min-Heap:** The value of the root node must be the **smallest** among all its children. This property must be recursively true for all subtrees.
2. **Max-Heap:** The value of the root node must be the **greatest** among all its children. This property must be recursively true for all subtrees.

---

### Core Operations & Internal Mechanics

Natively, heaps are built using a **Static or Dynamic Array** instead of raw pointers. Because it is a complete binary tree, its layout fits perfectly into an array without wasting memory space.

#### Map Tree Nodes to Array Indices
For any node stored at index `i` (using **0-based indexing**):
* **Left Child Index:** `2 * i + 1`
* **Right Child Index:** `2 * i + 2`
* **Parent Index:** `(i - 1) // 2`

#### Internal Balance Mechanics
* **Heapify-Up (Bubble-Up):** Used during insertions. If a newly added element violates the heap property (e.g., it is smaller than its parent in a Min-Heap), it swaps places with its parent. This repeats moving upwards until the property is restored.
* **Heapify-Down (Sift-Down):** Used during deletions. When the root is removed, the last element in the array replaces it. If this node violates the heap property, it swaps places with its smallest child (in a Min-Heap) or largest child (in a Max-Heap). This repeats moving downwards.

---

### Complexity Analysis

| Operation | Time Complexity | Space Complexity | Description |
| :--- | :--- | :--- | :--- |
| **Insert** | O(log N) | O(1) | Pushes element to end, then runs Heapify-Up. |
| **Extract Min/Max** | O(log N) | O(1) | Removes root, replaces it with last element, runs Heapify-Down. |
| **Peek Min/Max** | O(1) | O(1) | Simply returns the element at index `0`. |
| **Build Heap** | O(N) | O(1) | Converts an unsorted array into a heap in-place. |



# Heap and Priority Queue 

```python
# Build Min Heap (Heapify)

# Time Complexity: O(n) and Space Complexity: O(1)
import heapq

A = [-4, 2, -3, 5, 6, 10, 8, 12, 9, 7] # Binary Tree representation of the array
heapq.heapify(A)
print(A) # Heapified Array
```

```python
# Heap Push (Insert element into heap)
# Time complexity : O(log n) and Space Complexity: O(1)

heapq.heappush(A, -100)
A
```

```python
# Heap Pop (Extract minimum element from heap)
# Time complexity : O(log n) and Space Complexity: O(1)
min_element = heapq.heappop(A)
print("Min Heap : ", A)
print(min_element)
```

```python
# Heap sort (Sort the array using heap)
# Time complexity : O(n log n) and Space Complexity: O(1)
heapq.heapify(A)
sorted_array = [] 
while A:
    sorted_array.append(heapq.heappop(A))
    print(sorted_array, len(sorted_array))
```

```python
# Heap Push pop ( Insert and extract minimum element from heap)
# Time complexity : O(log n) and Space Complexity: O(1)

A = [2,4,-2,3, 0, 100]

heapq.heapify(A)
print("Min Heap : ", A)
min_element = heapq.heappushpop(A, 999)
print("Min element : ",min_element)
print("New Min Heap : ",A)
```

## Max heap 

```python
A = [-4, 2, -3, 5, 6, 10, 8, 12, 9, 7] # Binary Tree 

# In python heapq does only support min heap. So we can convert it to max heap by multiplying all the elements with -1.

for i in range(len(A)):
    A[i] = -A[i]

heapq.heapify(A)
A # Max Heap
```

```python
# To Get the largest from the max heap
# Time complexity : O(log n) and Space Complexity: O(1)

max_element = heapq.heappop(A) * (-1) # We multiply by -1 to get the original value
max_element
```

```python
# Heap push in max heap
# Time complexity : O(log n) and Space Complexity: O(1)

heapq.heappush(A, -999) # If want to insert a positive number we can multiply by -1 before inserting. 
A
```

```python
# Peak element in max heap
# Time complexity : O(1) and Space Complexity: O(1)
A[0]
```

```python
# Build Heap ( if need to go through it)
# Time complexity : O(n log n)

x = [10, 9, 11, 7, -3, -2, 19]

heap = []
for i in x:
    heapq.heappush(heap, i)

heap # min heap 
```

```python
# Putting tuples of items on the heap

y = [1,3,4,5,6,1,1,2,3,5,5,6,3,3,4,5,5]

from collections import Counter

counter = Counter(y)
print(counter)


heap =[]
for key,val in counter.items():
    heapq.heappush(heap, (val,key))

print(heap) # min heap of values containing the frequency
```
