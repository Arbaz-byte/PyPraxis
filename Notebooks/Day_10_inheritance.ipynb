# Inheritance 
Inheritance is a mechanism in object-oriented programming where one class can inherit the properties and methods of another class. This allows for code reuse and the creation of more specialized classes based on existing ones.


### 1. Single Inheritance

```python

# Parent Class

class Car:
    ##Constructor
    def __init__(self, Model_year, color ,engine_type='NAN'):
        self.Model_year = Model_year
        self.engine_type = engine_type
        self.color = color
    
    # Instance Method
    def start_engine(self):
        print("Engine started!! ")
    
    def car_info(self):
        print(f"Model Year: {self.Model_year} | Engine Type: {self.engine_type} | Color: {self.color}")


car1=Car(2026,'black',"V8")
car1.start_engine()
car1.car_info()

```

```python
# Child class 
class ElectricCar(Car):
    def __init__(self, Model_year, color ,engine_type='NAN', battery_capacity=0):
        super().__init__(Model_year, color ,engine_type='NAN')
        self.battery_capacity = battery_capacity
    
    def self_drive(self):
        print("Self Driving Mode Activated! ")
    
    def car_info(self):
        print(f"Model Year: {self.Model_year} | Engine Type: {self.engine_type} | Color: {self.color} | Battery Capacity: {self.battery_capacity}")
        
        

tesla = ElectricCar(2025,'black',engine_type='NAN' , battery_capacity=100 )
tesla.start_engine()
tesla.self_drive()
tesla.car_info()
```

### 2. Multiple Inheritance
In multiple inheritance, a child class can inherit from more than one parent class. This means that the child class will have all the attributes and methods of both parent classes.


```python
# Parent class 1
class Animal:
    def __init__(self, name):
        self.name = name
        
    def speak(self):
        pass    
    
# Parent Class 2:
class pet:
    def __init__(self, owner):
        self.owner = owner

# child class inheriting from both parent classes
class Dog(Animal,pet):
    
    def __init__(self, name, owner):
        Animal.__init__(self,name)
        pet.__init__(self,owner)
    
    def move(self):
        print("The dog is moving")

dog1 = Dog('Buddy','John')
print(dog1.name) # Output: Buddy
print(dog1.owner) # Output: John
dog1.move()
```
