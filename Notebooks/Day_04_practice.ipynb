# Function

- A function is a specific code that performs a specific task. 
- Function help in organizing code ,reusing code, and improving readability

```python

# Create function
def check_even(num):
    """ This function finds even or odd"""
    if num%2==0 :
        print("Even Number")
    else:
        print("Odd Number") # Not necessary to have return values 

# Function call 
check_even(12)
```

```python
# Function with multiple parameters 

def great_num(a,b,c):
    """ This function finds largest number among three """    
    if a>=b and a>=c :
        print("Largest :",a)
    elif b>=a and b>=c :
        print("Largest :",b)
    else:
        print("Largest :",c)        


# Test Cases:
great_num(1,2,2)
great_num(1,1,1)
great_num(4,5,9)
```

```python
# Default parameter 

def customer_rating(star_num=2):
    
    """ This function classify customer ratings 
    Note : Star_num [0,5] """
    
    if star_num == 5:
        print(f"Customer Rate your Drink as : {star_num} stars, Great Work!")
    elif star_num < 5 and star_num >= 2.5:
        print(f"Customer Rate your Drink as : {star_num} stars, Average Work!")  
    else:
        print("Customer Is not Satisfied with your work! ")    

# Test cases:
customer_rating(0)
customer_rating(4.5)
customer_rating(5)
```

```python
# Variable length Argument 

# 1. Positional Arguments

def prime(*args): 
    """ This function will take any length of number  and return prime numbers out of it"""
    lst = []
    for num in args:
        if num > 1: # Prime numbers are greater than 1
            for i in range(2,num):
                if num%i == 0: # check for other factors other than 1 and number itself
                    break
            else:
                lst.append(num)
    return lst            

prime(2,3,4,5,7,10,11,15,6,19)
```

```python

# 2. Keywords Arguments (key-value pairs -- dictionary)

def details(**kwargs):
    
    # Access like dictionary 
    for key,value in kwargs.items():
        print(f"{key} : {value}")
        


details(name="Arbaz", age=22, Gender="M") # IF the input is in key-value pairs all them referenced as **kwargs
```

```python

# 3. Mixed of positional and keyword arguments 

def mixed(*args, **kwargs):
    print("Positional Argument:")
    for val in args:
        print(val)
        
    print("Keyword Arguments : ")
    for key,val in kwargs.items():
        print(f"{key} : {val}")

mixed(1,2,3,4,name="Iron Man", arm= "Mark 42")            
```

## Function Examples 

### Example 1: Temperature Conversion

```python
def convert_temp(temp, unit):
    """ This function convert temperature b/w Celsius and Fahrenheit"""
    
    if unit == "C":
        return (9/5 * temp) + 32 # Celsius to Fahrenheit
    else:
        return ((temp-32)*5)/9 # Fahrenheit to Celsius
    
convert_temp(0,"C")  # O degree celsius is 32F  
```

### Example 2: Password Strength Check

```python
def eval_password(password):
    
    """" This function checks if the password is strong or not """
    
    if len(password)<8:
        return False
    elif not any(char.isdigit() for char in password):
        return False 
    elif not any(char.islower() for char in password):
        return False
    elif not any(char.isupper() for char in password):
        return False
    elif not any(char in '!@#$%^&*?' for char in password):
        return False
    return True

print("Is 'arbazkhan' a strong Password ? :",eval_password("arbazkhan"))
print("Is 'Arbazkhan67**' a strong Password ? :", eval_password("Arbazkhan67**"))
```

### Example 3 : Calculate the Total Cost of items 

```python

def cost_eval(cart):
    
    data = []
    for item in cart:
        each_cost = {
            'item' :item['name'],
            'cost' : item['price']*item['quantity']
        }
        data.append(each_cost)
    cost = sum(item['cost'] for item in data)
    return data,cost   


# Cart data 
cart = [
    
    {'name':'iphone', 'price':999,'quantity': 2},
    {'name':'cables', 'price':49,'quantity': 3},
    {'name':'charger', 'price':89,'quantity': 1}
]

data,cost = cost_eval(cart)
print("Cart Details : ", data)
print("Total cost : ", cost)
```

### Example 4 : Check if a string is palindrome 

```python
def check_palindrome(string):
    string = string.lower().replace(" ","")
    return string == string[::-1]

print("IS Arbaz a palindrome ? : ",check_palindrome("Arbaz"))

print("IS  'A man a plan a canal Panama' a palindrome ? : ",check_palindrome("A man a plan a canal Panama"))
```

### Example 5 : Calculate the factorial of a number using recursion

```python
def factorial(n):
    
    if n==0:
        return 1
    else:
        return n * factorial(n-1)

print(factorial(2))
print(factorial(1))
```
