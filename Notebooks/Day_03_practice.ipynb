# Tuples
- Tuples are ordered collections of items that are immutable. 
- They are similar to list, but their immutability makes them different.

```python
# Create a tuple

# 1.
empty_tuple = ()

# 2.
empty_tuple1 = tuple()

print(type(empty_tuple))
print(type(empty_tuple1))
```

```python
# list ---> tuple and tuple --> list
a = [1,2,3,4]
b = tuple(a)
c = list(b)

print("Tuple : ", b)
print("List : ",c)
```

```python
# tuple comprehension 
numbers = (i for i in range(1,11))
print(numbers) # gives the location in memory 

numbers1 = tuple([i for i in range(1,11)])
print(numbers1) # Type casting list into tuple


# Accessing tuple elements (Similar to list)
print(numbers1[1])

# Slicing tuple
num_tuple = tuple([i for i in range(1,21)])
print(num_tuple[::-1]) # Reversing tuple
print(num_tuple[::2])
```

```python
# Tuples operations

t1 = (1,2,3,4,5,6)
t2 = ('a',True,3.12)

# Tuple Concatenation 
print(t1+t2)

# Tuple multiplication
print(t1*2) # Concatenate same tuple two times
# print(t1 * t2) -- invalid operation (Type error)



```

```python
# Immutable Nature of tuple 
print("Can't change the elements of tuple -- > Immutable")

tup = (1,2,3,4,5,6)
print("Before : ",tup)

'''
tup[0]='arbaz'
print("After : ",tup) ''' # TypeError

# To change an element of tuple convert ---> list ---> change element --->  tuple

# Mutable Nature of list
print("Can change the element in the list -- > Mutable")
lst = [2,3,4,5,6]
print("Before : ",lst)

lst[0]='arbaz'
print("After : ",lst)
```

```python
# Tuple methods

## 1. Count frequency of element in tuple 
tup1 = tuple([j for j in range(5) for i in range(1,j+1)])
print(tup1.count(3)) # counts the frequency

## 2. Get index of element in tuple 
print(tup1.index(2)) # return first occur element
```

```python
# Packing tuple
packed_tuple = 1, 10, 'macbook'
print(packed_tuple)

# Unpacking tuple
a,b,c = packed_tuple #Need same no of variable as elements (problem)
print(a,b,c)

# Unpacking with * (solution)
num = tuple([i for i in range(1,11)])

first, *middle, last = num # left over elements would be there in middle variable
print(first)
print(middle) # return as list 
print(last)

```

```python
# Nested tuples

tup = ( (1,3,5,7), (2,4,6,8), (True, False))
print(tup[0])
print(tup[0][1])

# Iterating over tuple
for elem in tup:
    for item in elem:
        print(item, end =' ') # end = ' ' for space b/e elements 
    print() # for next line 
```
