# The Lambda Function

- Lambda function are small anonymous function(without a name) defined using **Lambda** keyword.
- It can have any number of arguments but only one expression.
- Commonly used for short operations or arguments to higher order functions.

```python
"""
Syntax :
lambda variable: expression
"""

add = lambda a,b : a+b

print(type(add))
print(add(2,4))
```

```python
# Check even 
even  = lambda num:num%2 == 0
print("Is even? : ",even(2))
```

# The Map() function

- The map() function applies given functions to all items in an input list(or any other iterable) and returns a map object(an iterator).
- Particularly useful for transforming data in a list comprehensively 


```python
# Square of the number 
def square(x):
    return x**2

""" To calculate the square of list of number for loop is used but map() can do it easily 

    map( function,  iterable) --> gives map at memory convert it into list 

"""

numbers = [1,2,3,4,5,6,7]
print("Map at : ", map(square, numbers))
print("List of square : ", list(map(square, numbers)) ) 
```

```python
## Lambda function with map
numbers = [10, 20, 30, 40]

list(map(lambda x:x**2, numbers))
```

```python
## Map Multiple iterable 

num1 = [1,2,3]
num2 = [3,2,1]

add_num = list(map(lambda x,y: x+y , num1,num2))
print(add_num)
```

```python
## Using Map() to convert list of strings to integers

str_num = [ '1', '2', '3', '4', '5', 3.3]
int_num = list(map(int, str_num)) # Type cast 

print(int_num)
```

```python
## using in-built class 

words = ['apple', 'banana', 'cherry']
name = ['Arbaz', 'ThOr', 'IRON Man']

upper_case = list(map(str.upper, words))
lower_case = list(map(str.lower, name))
print(upper_case)
print(lower_case)
```

```python
def get_name(person):
    return person['name']

people = [
    {'name': 'Thor', 'age': 150},
    {'name': 'Iron man', 'age': 55},
    {'name': 'Dr. Doom', 'age': 'NAN'}
]

name_info = list(map(get_name, people))
age_info = list(map(lambda x:x['age'], people))


print(name_info)
print(age_info)
```

# The Filter() Function

- The filter() function construct and iterable from elements of an iterable for which a function returns true.
- It is used to filter out items from list (or any iterable) based on a condition


```python
""" 
filter() --> extract elements that satisfy the condition 

syntax:
    filter(function, iterable)

"""

lst_num = [1,3,4,5,6,7,8,9,10]

even = lambda x:x%2 ==0

filter(even, lst_num) # --> memory location

print(list(filter(even, lst_num))) # apply filter and extract elements that satisfy condition

```

```python
## FIlter function with lambda function and multiple condition

lst_num = [1,3,4,5,6,7,8,9,10]

lst_extract = list(filter(lambda x:x>5 and x%2==0, lst_num))
print("Even and greater than 5 : ", lst_extract)
```

```python
## Filter() to check age condition
people = [
    {'name': 'Thor', 'age': 150},
    {'name': 'Iron man', 'age': 55},
    {'name': 'Dr. Doom', 'age': 50},
    {'name': 'spider man', 'age': 26}
    
]

age_cond = list(filter(lambda x:x['age']>100 , people))
print(age_cond)
```
