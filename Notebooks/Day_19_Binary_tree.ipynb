# **Introduction to Binary Trees & Pre-Order Traversal (DFS)**

A **Binary Tree** is a hierarchical data structure in which each node has at most two children, referred to as the left child and the right child. **Pre-Order Traversal** explores the tree by visiting the root node first, then moving to the left subtree, and finally the right subtree.

*   **Order:** `Root ➔ Left ➔ Right`

#### **1. Recursive Method**
*   **Algorithm:**
    1. Check if the current node is `NULL`. If yes, return.
    2. Process/Print the current node's data.
    3. Recursively call Pre-order on the left child.
    4. Recursively call Pre-order on the right child.

#### **2. Iterative Method**
*   **Algorithm:**
    1. Initialize an empty stack and push the `root` node onto it.
    2. Loop while the stack is not empty:
       * Pop the top node and process it.
       * Push the **right child** of the popped node onto the stack (if it exists).
       * Push the **left child** of the popped node onto the stack (if it exists).

#### **Complexity Analysis**
*   **Time Complexity:** 
    *   **Best Case:** $\mathcal{O}(N)$ — Every node is visited exactly once.
    *   **Worst Case:** $\mathcal{O}(N)$ — Every node is visited exactly once.
*   **Space Complexity:**
    *   **Best Case (Balanced Tree):** $\mathcal{O}(\log N)$ — Stack space is proportional to tree height.
    *   **Worst Case (Skewed Tree):** $\mathcal{O}(N)$ — Tree height equals the total number of nodes.


```python
class TreeNodes:
    def __init__(self, val, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
    
    def __str__(self):
        return str(self.val)



# Example Binary Tree
""" 
                1
               / \
              2   3
             / \ / \
            9  5 10
"""

root = TreeNodes(1)
root.left = TreeNodes(2)
root.right = TreeNodes(3)
root.left.left = TreeNodes(9)
root.left.right = TreeNodes(5)
root.right.left = TreeNodes(10)


print(""" 
TreeNode(1, TreeNode(2, TreeNode(9, None, None), TreeNode(5, None, None)), TreeNode(3, TreeNode(10, None, None), None))
""")
```

### **1. In-Order Traversal (DFS)**

**In-Order Traversal** processes the left subtree first, visits the root node next, and then moves to the right subtree. In a Binary Search Tree, this traversal yields elements in sorted ascending order.

*   **Order:** `Left ➔ Root ➔ Right`

#### **1. Recursive Method**
*   **Algorithm:**
    1. Check if the current node is `NULL`. If yes, return.
    2. Recursively call In-order on the left child.
    3. Process/Print the current node's data.
    4. Recursively call In-order on the right child.

#### **2. Iterative Method**
*   **Algorithm:**
    1. Initialize an empty stack and set a pointer `current` to the `root`.
    2. Loop while `current` is not `NULL` or the stack is not empty:
       * Push all left nodes of the `current` node onto the stack until reaching the leftmost node.
       * Pop the top node from the stack and process it.
       * Set `current` to the **right child** of the popped node.

#### **Complexity Analysis**
*   **Time Complexity:** 
    *   **Best Case:** $\mathcal{O}(N)$
    *   **Worst Case:** $\mathcal{O}(N)$
*   **Space Complexity:**
    *   **Best Case (Balanced Tree):** $\mathcal{O}(\log N)$ — Maximum stack depth matches the tree height.
    *   **Worst Case (Skewed Tree):** $\mathcal{O}(N)$ — Stack holds all nodes in a single line.


```python
# Recursive In-order Traversal (DFS) 
# Time complexity : O(n) where n is the number of nodes in the tree. 
# Space complexity : O(h) where h is the height of the tree. in worse case h =n (linked list)

def inOrder(node):
    if not node:
        return 
    
    inOrder(node.left)
    print(node)
    inOrder(node.right)

inOrder(root)
```

```python
# Iterative in-order Traversal (DFS) 
# Time complexity : O(n) where n is the number of nodes in the tree. 
# Space complexity : O(h) where h is the height of the tree. in worse case h =n (linked list)


def inOrder_iterative(node):
    stack = []
    current = node
    
    
    while current is not None or stack:
        while current is not None:
            stack.append(current)
            current = current.left
        
        current = stack.pop()
        print(current)
        current = current.right


inOrder_iterative(root)
```

### **2. Post-Order Traversal (DFS)**

**Post-Order Traversal** completely processes both the left and right subtrees before visiting their parent root node.

*   **Order:** `Left ➔ Right ➔ Root`

#### **1. Recursive Method**
*   **Algorithm:**
    1. Check if the current node is `NULL`. If yes, return.
    2. Recursively call Post-order on the left child.
    3. Recursively call Post-order on the right child.
    4. Process/Print the current node's data.

#### **2. Iterative Method (Using Two Stacks)**
*   **Algorithm:**
    1. Initialize two stacks: `stack1` and `stack2`. Push `root` to `stack1`.
    2. Loop while `stack1` is not empty:
       * Pop a node from `stack1` and push it to `stack2`.
       * Push the node's **left child** to `stack1` (if it exists).
       * Push the node's **right child** to `stack1` (if it exists).
    3. Pop and print all elements from `stack2` sequentially to get the final Post-order sequence.

#### **Complexity Analysis**
*   **Time Complexity:** 
    *   **Best Case:** $\mathcal{O}(N)$
    *   **Worst Case:** $\mathcal{O}(N)$
*   **Space Complexity:**
    *   **Best Case (Balanced Tree):** $\mathcal{O}(N)$ — Due to storing nodes in the secondary stack.
    *   **Worst Case (Skewed Tree):** $\mathcal{O}(N)$


```python
# Recursive Post-order traversal (DFS) 
# Time complexity : O(n) where n is the number of nodes in the tree. 
# Space complexity : O(h) where h is the height of the tree. in worse case h = n (linked list)

def postOrder(node):
    if not node:
        return 
    postOrder(node.left)
    postOrder(node.right)
    print(node)


postOrder(root)
```

```python
# Iterative Pre-order Traversal (DFS) 
# Time complexity : O(n) where n is the number of nodes in the tree. 
# Space complexity : O(h) where h is the height of the tree. in worse case h =n (linked list)

def pre_order_iterative(node):
    stack = [node]
    
    while stack:
        node = stack.pop()
        print(node)
        
        if node.right: stack.append(node.right)
        if node.left: stack.append(node.left)
    

pre_order_iterative(root)
```

### **3. Level-Order Traversal (BFS)**

**Level-Order Traversal** visits nodes layer by layer, starting from the root (Level 0) and moving from left to right across each subsequent depth level.

*   **Order:** `Level 0 ➔ Level 1 ➔ Level 2 ...`

#### **1. Iterative Method (Standard Queue Approach)**
*   **Algorithm:**
    1. Initialize an empty queue and push the `root` node.
    2. Loop while the queue is not empty:
       * Dequeue the front node and process/print it.
       * Enqueue the **left child** if it exists.
       * Enqueue the **right child** if it exists.

#### **2. Recursive Method (Simulated Levels)**
*   **Algorithm:**
    1. Calculate the total height of the tree.
    2. Loop a counter variable from level 1 up to `height`:
       * Call a helper function `printCurrentLevel(root, level)` that decrements the targeted level step-by-step until it reaches 1, then outputs the node data.

#### **Complexity Analysis**
*   **Time Complexity:** 
    *   **Best Case:** $\mathcal{O}(N)$ — Every node must be touched once.
    *   **Worst Case:** $\mathcal{O}(N)$
*   **Space Complexity:**
    *   **Best Case (Skewed Tree):** $\mathcal{O}(1)$ — The queue holds at most 1 node at any given moment.
    *   **Worst Case (Perfect Balanced Tree):** $\mathcal{O}(N)$ — The widest leaf level holds $\lceil N/2 \rceil$ nodes, requiring the queue to store up to half the tree at its peak.


```python
# Level Order Traversal (BFS) 
# Time complexity : O(n) where n is the number of nodes in the tree. 
# Space complexity : O(h) where h is the height of the tree. in worse case h =  n (linked list)

from collections import deque

def levelOrder(node):
    q = deque()
    q.append(node)
    
    while q:
        node = q.popleft()
        print(node)
        
        if node.left : q.append(node.left)
        if node.right : q.append(node.right)


levelOrder(root)
```

```python
# check if value exist (DFS)?
# Time complexity : O(n) where n is the number of nodes in the tree. 
# Space complexity : O(h) where h is the height of the tree. in worse case h = n (linked list)


def search_node(node, target):
    
    if not node:
        return False
    
    if node.val == target:
        return True
    
    return search_node(node.left, target) or search_node(node.right, target)


search_node(root, 100)
```

### **4. Binary Search Tree (BST) Theory**

A **Binary Search Tree (BST)** is a variant of a binary tree that maintains a strict relative order among its keys to allow for rapid lookups, insertions, and removals.

#### **The Core BST Property**
For every given node `X` in the tree:
*   All keys in the **left subtree** of `X` must be strictly **less than** the key of `X`.
*   All keys in the **right subtree** of `X` must be strictly **greater than** the key of `X`.

#### **Core Operations**
1.  **Search:** Compare the target value with the current node. If it is smaller, branch left; if larger, branch right. Repeat until found or a `NULL` leaf boundary is encountered.
2.  **Insertion:** Follows the standard search path down the tree until a structural `NULL` gap is reached, appending the new leaf node there.
3.  **Deletion:** Involves three structural cases: deleting a node with zero children (leaf), one child (bypass node), or two children (replace node with its in-order successor or predecessor).

#### **Complexity Analysis**

| Operation | Best Case (Balanced Tree) | Worst Case (Skewed Tree) |
| :--- | :--- | :--- |
| **Search** | $\mathcal{O}(\log N)$ | $\mathcal{O}(N)$ |
| **Insertion** | $\mathcal{O}(\log N)$ | $\mathcal{O}(N)$ |
| **Deletion** | $\mathcal{O}(\log N)$ | $\mathcal{O}(N)$ |
| **Space Complexity** | $\mathcal{O}(\log N)$ *(Call Stack)* | $\mathcal{O}(N)$ *(Call Stack)* |

*Note: Self-Balancing Binary Search Trees (such as AVL Trees or Red-Black Trees) dynamically restructure themselves during adjustments to guarantee $\mathcal{O}(\log N)$ operational performance limits.*

```python
""" 
                1
               /  \
              1    8
             / \  /  \
            -1  3  7   9
"""

root = TreeNodes(1)
root.left = TreeNodes(1)
root.right = TreeNodes(8)
root.left.left = TreeNodes(-1)
root.left.right = TreeNodes(3)
root.right.left = TreeNodes(7)
root.right.right = TreeNodes(9)
```

```python
# Time Complexity : O(log N) if tree is balanced. In worst case, it will be O(N). 
# Space complexity : O(h) where h is the height of the tree. in worse case h = n (linked list)

def search_bst(node, target):
    
    if not node:
        return False
    
    if node.val == target:
        return True
    
    if node.val > target:
        return search_bst(node.left, target)
    elif node.val < target:
        return search_bst(node.right, target)

search_bst(root, 8)
```
