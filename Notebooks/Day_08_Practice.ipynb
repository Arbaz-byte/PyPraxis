# Understanding Exceptions

- It is allows us to handle errors gracefully and take corrective actions without stopping the execution of the program.
- When an error occurs, Python raises an exception.


## What Are Exceptions

- An exception is a runtime error that occurs when the code violates certain rules or constraints.
- It can be caused by various factors such as invalid input, division by zero, file not found, etc.

## Types of Exceptions

- **SyntaxError**: This occurs when there is a syntax error in the code
- **NameError**: This occurs when a variable is used before it has been defined.
- **TypeError**: This occurs when an operation or function is
                executed on an object of inappropriate type.
- **ValueError**: This occurs when a function receives an argument that is not within its valid range.
- **IndexError**: This occurs when trying to access an index in a sequence that does not exist.
- **KeyError**: This occurs when trying to access a key in a dictionary that does not exist.
- **FileNotFoundError**: This occurs when trying to open a file that does not exist.
- **ZeroDivisionError**: This occurs when dividing by zero.


## Handling Exceptions

- We can handle exceptions using the `try` and `except` blocks.
- The `try` block contains
- The `except` block catches the exception and handles it.
- We can also use multiple `except` blocks to catch different types of exceptions.
- We can also use a `finally` block that executes regardless of whether an exception was raised or not.





```python
# Exception try, except block

# Method 1:
try:
    a=b
except:
    print("The variable has not been assigned")
    

# Method 2:
try:
    a=b
except NameError as e:
    print(e)    
```

```python
# Zerodivision error 

try:
    result = 1 /0
except ZeroDivisionError as e:
    print("ZeroDivision Error: ",e)
    print("Denominator Should be Non-Zero Number!")


```

```python
# Exception class -- Handles all exception

try:
    a=b
    result =1/0
except Exception as e:
    print("Exception Occured : ",e)    #Caught Only first Exception
        
```

```python
# Multi Exception Handling

try :
    num = int(input("Enter a Number : "))
    
except ValueError as e :
    print("Value Error : ",e)
except ZeroDivisionError :
    print("Denominator Should be Non-Zero Number!")
except Exception as e :
    print("Exception Occured : ",e)    
    

```

```python
# Exception Handling with Else 

try :
    num = int(input("Enter a Number : "))
    result = 10/num
    
except Exception as e :
    print("Exception Occured : ",e)
else:
    print(f"The result is : {result}")   # Only executed when error is not there
```

```python
# Exception Handling with Finally

try :
    num = int(input("Enter a Number : "))
    result = 10/num
except Exception as e :
    print("Exception Occured : ",e)
else:
    print(f"The result is : {result}")        
finally :
    print("Execution Complete!")  # Always Executed 
```

```python
# File handling and exception handling

# Method 1:
try:
    file=open('./package/test/test3.txt','r')
except FileNotFoundError as e :
    print("File Not Found Error : ",e) 
else:
    print(file.read())
finally :
    file.close()  # Always Executed 
```

# Custom Exception (Raise and Throw an Exception)

- custom exception is a user-defined error that can be raised when an error occurs in the program.
- it is defined by the user using the raise keyword followed by an exception class and an optional message that describes the error.
- it can be raised anywhere in the program using the raise keyword followed by an exception class and an optional message that describes the error.
- it can be caught using a try-except block and the except clause can handle the exception by printing an error message or taking some other action. 





```python
# Custom Exception Example:

class error(Exception):
    pass


class InvalidDOB (error):
    pass

age = int(input("Enter your Age : "))
try :
    if age>=20 and age<=30:
        print("You are Eligible for the Job !")
    else :
        raise InvalidDOB
    
except InvalidDOB:
    print("Your DOB is not Eligible for the Job !")
except ValueError:
    print("Invalid Input! Please enter a valid number.")
    
finally:
    print("Thank you for using the program !")
```
